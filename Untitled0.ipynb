```python
# ============================================================
# ROBOT ARM POSITION CALCULATOR - BACKEND
# 2-Link 2D Robot Arm | Inverse + Forward Kinematics
# ============================================================

import numpy as np
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation
from IPython.display import HTML, display

# ============================================================
# 1. USER INPUTS - IKKADA VALUES CHANGE CHEYYACHU
# ============================================================

L1 = 5.0          # First link length
L2 = 3.0          # Second link length

target_x = 6.0    # Desired X position
target_y = 2.0    # Desired Y position

# "up" = Elbow Up
# "down" = Elbow Down
elbow_mode = "up"


# ============================================================
# 2. CHECK TARGET REACHABILITY
# ============================================================

distance = np.sqrt(target_x**2 + target_y**2)

max_reach = L1 + L2
min_reach = abs(L1 - L2)

print("=" * 55)
print("        ROBOT ARM POSITION CALCULATOR")
print("=" * 55)

print(f"\nLink 1 Length : {L1}")
print(f"Link 2 Length : {L2}")
print(f"Target X      : {target_x}")
print(f"Target Y      : {target_y}")
print(f"Target Distance: {distance:.3f}")

if distance > max_reach or distance < min_reach:
    print("\n❌ TARGET IS NOT REACHABLE")
    print(f"Reachable range: {min_reach:.2f} to {max_reach:.2f}")

else:

    print("\n✅ TARGET IS REACHABLE")

    # ========================================================
    # 3. INVERSE KINEMATICS
    # ========================================================

    cos_theta2 = (
        target_x**2 +
        target_y**2 -
        L1**2 -
        L2**2
    ) / (2 * L1 * L2)

    # Avoid floating-point errors
    cos_theta2 = np.clip(cos_theta2, -1, 1)

    theta2 = np.arccos(cos_theta2)

    # Elbow Up / Down
    if elbow_mode.lower() == "down":
        theta2 = -theta2

    theta1 = (
        np.arctan2(target_y, target_x)
        -
        np.arctan2(
            L2 * np.sin(theta2),
            L1 + L2 * np.cos(theta2)
        )
    )

    # Convert radians to degrees
    theta1_deg = np.degrees(theta1)
    theta2_deg = np.degrees(theta2)

    print("\n---------------- INVERSE KINEMATICS ----------------")

    print(f"Theta 1 = {theta1_deg:.3f}°")
    print(f"Theta 2 = {theta2_deg:.3f}°")


    # ========================================================
    # 4. FORWARD KINEMATICS
    # ========================================================

    joint1_x = L1 * np.cos(theta1)
    joint1_y = L1 * np.sin(theta1)

    end_x = (
        L1 * np.cos(theta1)
        +
        L2 * np.cos(theta1 + theta2)
    )

    end_y = (
        L1 * np.sin(theta1)
        +
        L2 * np.sin(theta1 + theta2)
    )

    print("\n---------------- FORWARD KINEMATICS ----------------")

    print(f"Calculated X = {end_x:.3f}")
    print(f"Calculated Y = {end_y:.3f}")


    # ========================================================
    # 5. POSITION ERROR
    # ========================================================

    error = np.sqrt(
        (target_x - end_x)**2 +
        (target_y - end_y)**2
    )

    print("\n---------------- VERIFICATION ----------------")

    print(f"Target Position     : ({target_x:.3f}, {target_y:.3f})")
    print(f"Calculated Position : ({end_x:.3f}, {end_y:.3f})")
    print(f"Position Error      : {error:.8f}")

    if error < 0.0001:
        print("✅ POSITION VERIFIED")
    else:
        print("⚠️ SMALL NUMERICAL ERROR")


    # ========================================================
    # 6. ROTATION MATRICES
    # ========================================================

    R1 = np.array([
        [np.cos(theta1), -np.sin(theta1)],
        [np.sin(theta1),  np.cos(theta1)]
    ])

    R2 = np.array([
        [np.cos(theta2), -np.sin(theta2)],
        [np.sin(theta2),  np.cos(theta2)]
    ])

    R_total = R1 @ R2

    print("\n---------------- ROTATION MATRICES ----------------")

    print("\nR1 =")
    print(np.round(R1, 4))

    print("\nR2 =")
    print(np.round(R2, 4))

    print("\nTotal Rotation Matrix = R1 × R2")
    print(np.round(R_total, 4))


    # ========================================================
    # 7. RESULT TABLE
    # ========================================================

    print("\n---------------- FINAL RESULTS ----------------")

    print(f"""
    Link 1 Length     : {L1}
    Link 2 Length     : {L2}

    Target X          : {target_x}
    Target Y          : {target_y}

    Joint Angle θ1    : {theta1_deg:.3f}°
    Joint Angle θ2    : {theta2_deg:.3f}°

    Final X           : {end_x:.3f}
    Final Y           : {end_y:.3f}

    Position Error    : {error:.8f}
    """)


    # ========================================================
    # 8. ROBOT ARM STATIC VISUALIZATION
    # ========================================================

    plt.figure(figsize=(8, 8))

    # Base → Joint 1
    plt.plot(
        [0, joint1_x],
        [0, joint1_y],
        marker="o",
        linewidth=5
    )

    # Joint 1 → End Effector
    plt.plot(
        [joint1_x, end_x],
        [joint1_y, end_y],
        marker="o",
        linewidth=5
    )

    # Target
    plt.scatter(
        target_x,
        target_y,
        s=150,
        marker="x",
        label="Target"
    )

    # Base
    plt.scatter(0, 0, s=120, label="Base")

    plt.axhline(0, linewidth=1)
    plt.axvline(0, linewidth=1)

    plt.xlim(-(L1 + L2 + 1), L1 + L2 + 1)
    plt.ylim(-(L1 + L2 + 1), L1 + L2 + 1)

    plt.xlabel("X Position")
    plt.ylabel("Y Position")
    plt.title("2-Link Robot Arm - Inverse Kinematics")

    plt.grid(True)
    plt.legend()
    plt.axis("equal")

    plt.show()


    # ========================================================
    # 9. ANIMATION
    # ========================================================

    fig, ax = plt.subplots(figsize=(7, 7))

    ax.set_xlim(-(L1 + L2 + 1), L1 + L2 + 1)
    ax.set_ylim(-(L1 + L2 + 1), L1 + L2 + 1)

    ax.set_aspect("equal")
    ax.grid(True)

    ax.set_xlabel("X Position")
    ax.set_ylabel("Y Position")

    ax.set_title("Robot Arm Movement")

    # Target point
    ax.scatter(
        target_x,
        target_y,
        marker="x",
        s=150
    )

    arm_line, = ax.plot([], [], marker="o", linewidth=5)

    def update(frame):

        # Smooth movement from 0 → final angles
        current_theta1 = theta1 * frame / 100
        current_theta2 = theta2 * frame / 100

        x1 = L1 * np.cos(current_theta1)
        y1 = L1 * np.sin(current_theta1)

        x2 = (
            x1 +
            L2 * np.cos(current_theta1 + current_theta2)
        )

        y2 = (
            y1 +
            L2 * np.sin(current_theta1 + current_theta2)
        )

        arm_line.set_data(
            [0, x1, x2],
            [0, y1, y2]
        )

        return arm_line,

    animation = FuncAnimation(
        fig,
        update,
        frames=101,
        interval=30,
        blit=True
    )

    plt.close(fig)

    display(HTML(animation.to_jshtml()))


# ============================================================
# 10. BENCHMARK DATASET
# ============================================================

print("\n========================================================")
print("                 BENCHMARK DATASET")
print("========================================================")

benchmark = [
    (5, 3, 6, 2),
    (4, 3, 5, 2),
    (6, 2, 4, 3),
    (5, 4, 3, 5)
]

for i, (a, b, x, y) in enumerate(benchmark, 1):

    d = np.sqrt(x**2 + y**2)

    reachable = (
        abs(a - b) <= d <= a + b
    )

    print(
        f"Test {i}: "
        f"L1={a}, L2={b}, "
        f"Target=({x},{y}) → "
        f"{'Reachable ✅' if reachable else 'Not Reachable ❌'}"
    )

print("\n✅ BACKEND EXECUTION COMPLETED")
```

```python
from IPython.display import HTML, display

html_code = r'''
<!DOCTYPE html>
<html>
<head>
<meta charset="UTF-8">

<style>
* {
    box-sizing: border-box;
}

body {
    margin: 0;
    font-family: Arial, sans-serif;
    background: #080d1d;
    color: white;
}

.container {
    max-width: 1200px;
    margin: auto;
    padding: 25px;
}

.header {
    text-align: center;
    padding: 20px;
    background: #111936;
    border-radius: 20px;
    margin-bottom: 20px;
}

.header h1 {
    margin: 0;
    font-size: 32px;
}

.header p {
    color: #aeb8d6;
}

.main {
    display: grid;
    grid-template-columns: 320px 1fr;
    gap: 20px;
}

.panel,
.visual {
    background: #111936;
    border: 1px solid #29365e;
    border-radius: 20px;
    padding: 20px;
}

label {
    display: block;
    margin-top: 15px;
    margin-bottom: 7px;
    color: #bbc5df;
}

input,
select {
    width: 100%;
    padding: 12px;
    background: #080d1d;
    color: white;
    border: 1px solid #394b7d;
    border-radius: 10px;
    font-size: 16px;
}

button {
    width: 100%;
    padding: 14px;
    margin-top: 18px;
    border: 0;
    border-radius: 10px;
    background: #4d7cff;
    color: white;
    font-size: 16px;
    font-weight: bold;
    cursor: pointer;
}

button:hover {
    background: #315fe0;
}

.reset {
    background: #273250;
}

canvas {
    width: 100%;
    height: 500px;
    background: #080d1d;
    border-radius: 15px;
}

.results {
    display: grid;
    grid-template-columns: repeat(3, 1fr);
    gap: 12px;
    margin-top: 15px;
}

.card {
    background: #080d1d;
    border: 1px solid #29365e;
    border-radius: 12px;
    padding: 15px;
    text-align: center;
}

.card small {
    color: #9da9c9;
}

.card h2 {
    margin: 8px 0 0;
}

.status {
    margin-top: 15px;
    padding: 12px;
    text-align: center;
    border-radius: 10px;
    background: #132b22;
    color: #62e6a1;
}

.formula {
    margin-top: 15px;
    padding: 18px;
    background: #080d1d;
    border-radius: 12px;
    line-height: 1.8;
    color: #bbc5df;
}

@media(max-width:800px) {
    .main {
        grid-template-columns: 1fr;
    }

    .results {
        grid-template-columns: 1fr;
    }
}
</style>
</head>

<body>

<div class="container">

<div class="header">
    <h1>🤖 Robot Arm Position Calculator</h1>
    <p>2-Link 2D Robot Arm • Inverse Kinematics • Mathematics</p>
</div>

<div class="main">

<div class="panel">

<h2>⚙️ Robot Inputs</h2>

<label>Link 1 Length</label>
<input id="L1" type="number" value="5" step="0.1">

<label>Link 2 Length</label>
<input id="L2" type="number" value="3" step="0.1">

<label>Target X</label>
<input id="targetX" type="number" value="6" step="0.1">

<label>Target Y</label>
<input id="targetY" type="number" value="2" step="0.1">

<label>IK Solution</label>

<select id="mode">
    <option value="up">Elbow Up</option>
    <option value="down">Elbow Down</option>
</select>

<button onclick="calculate()">
    🚀 Calculate Robot Position
</button>

<button class="reset" onclick="resetRobot()">
    🔄 Reset
</button>

</div>


<div class="visual">

<canvas id="canvas"></canvas>

<div class="results">

<div class="card">
    <small>JOINT ANGLE θ1</small>
    <h2 id="theta1">--</h2>
</div>

<div class="card">
    <small>JOINT ANGLE θ2</small>
    <h2 id="theta2">--</h2>
</div>

<div class="card">
    <small>POSITION ERROR</small>
    <h2 id="error">--</h2>
</div>

</div>

<div class="status" id="status">
    Enter values and calculate
</div>

<div class="formula">

<b>🧮 Mathematical Model</b>

<br><br>

x = L₁ cos(θ₁) + L₂ cos(θ₁ + θ₂)

<br>

y = L₁ sin(θ₁) + L₂ sin(θ₁ + θ₂)

<br><br>

cos(θ₂) =
(x² + y² − L₁² − L₂²) / (2L₁L₂)

</div>

</div>

</div>
</div>


<script>

const canvas = document.getElementById("canvas");
const ctx = canvas.getContext("2d");

let L1 = 5;
let L2 = 3;

let targetX = 6;
let targetY = 2;

let currentT1 = 0;
let currentT2 = 0;

let finalT1 = 0;
let finalT2 = 0;

let animation;


function resizeCanvas() {

    canvas.width = canvas.clientWidth;
    canvas.height = canvas.clientHeight;

    drawRobot(currentT1, currentT2);
}

window.addEventListener("resize", resizeCanvas);

resizeCanvas();


function calculate() {

    L1 = parseFloat(document.getElementById("L1").value);
    L2 = parseFloat(document.getElementById("L2").value);

    targetX = parseFloat(
        document.getElementById("targetX").value
    );

    targetY = parseFloat(
        document.getElementById("targetY").value
    );

    if (L1 <= 0 || L2 <= 0) {

        alert("Link lengths must be greater than 0.");

        return;
    }


    const distance = Math.sqrt(
        targetX * targetX +
        targetY * targetY
    );


    if (
        distance > L1 + L2 ||
        distance < Math.abs(L1 - L2)
    ) {

        document.getElementById("status").innerText =
            "❌ Target is unreachable";

        return;
    }


    /* Inverse Kinematics */

    let cosT2 =
        (
            targetX * targetX +
            targetY * targetY -
            L1 * L1 -
            L2 * L2
        )
        /
        (2 * L1 * L2);


    cosT2 = Math.max(
        -1,
        Math.min(1, cosT2)
    );


    let t2 = Math.acos(cosT2);


    const mode =
        document.getElementById("mode").value;


    if (mode === "down") {
        t2 = -t2;
    }


    let t1 =
        Math.atan2(targetY, targetX)
        -
        Math.atan2(
            L2 * Math.sin(t2),
            L1 + L2 * Math.cos(t2)
        );


    finalT1 = t1;
    finalT2 = t2;


    /* Forward Kinematics */

    const calculatedX =
        L1 * Math.cos(t1) +
        L2 * Math.cos(t1 + t2);


    const calculatedY =
        L1 * Math.sin(t1) +
        L2 * Math.sin(t1 + t2);


    /* Error */

    const error =
        Math.sqrt(
            Math.pow(calculatedX - targetX, 2) +
            Math.pow(calculatedY - targetY, 2)
        );


    document.getElementById("theta1").innerText =
        (t1 * 180 / Math.PI).toFixed(2) + "°";


    document.getElementById("theta2").innerText =
        (t2 * 180 / Math.PI).toFixed(2) + "°";


    document.getElementById("error").innerText =
        error.toFixed(6);


    document.getElementById("status").innerText =
        "✅ Target reached successfully";


    animate();
}


function drawRobot(t1, t2) {

    ctx.clearRect(
        0,
        0,
        canvas.width,
        canvas.height
    );


    const cx = canvas.width / 2;
    const cy = canvas.height / 2;


    const scale =
        Math.min(canvas.width, canvas.height) /
        (2 * (L1 + L2 + 1));


    const x1 =
        cx +
        L1 * scale * Math.cos(t1);


    const y1 =
        cy -
        L1 * scale * Math.sin(t1);


    const x2 =
        x1 +
        L2 * scale *
        Math.cos(t1 + t2);


    const y2 =
        y1 -
        L2 * scale *
        Math.sin(t1 + t2);


    /* Grid */

    ctx.strokeStyle = "#1d2744";
    ctx.lineWidth = 1;


    for (let x = 0; x < canvas.width; x += 40) {

        ctx.beginPath();
        ctx.moveTo(x, 0);
        ctx.lineTo(x, canvas.height);
        ctx.stroke();
    }


    for (let y = 0; y < canvas.height; y += 40) {

        ctx.beginPath();
        ctx.moveTo(0, y);
        ctx.lineTo(canvas.width, y);
        ctx.stroke();
    }


    /* Target */

    const tx =
        cx + targetX * scale;

    const ty =
        cy - targetY * scale;


    ctx.beginPath();

    ctx.arc(
        tx,
        ty,
        9,
        0,
        2 * Math.PI
    );

    ctx.strokeStyle = "#ff5577";
    ctx.lineWidth = 3;
    ctx.stroke();


    /* Link 1 */

    ctx.beginPath();

    ctx.moveTo(cx, cy);
    ctx.lineTo(x1, y1);

    ctx.strokeStyle = "#4d7cff";
    ctx.lineWidth = 14;
    ctx.lineCap = "round";

    ctx.stroke();


    /* Link 2 */

    ctx.beginPath();

    ctx.moveTo(x1, y1);
    ctx.lineTo(x2, y2);

    ctx.strokeStyle = "#62e6a1";
    ctx.lineWidth = 14;

    ctx.stroke();


    /* Base */

    ctx.beginPath();

    ctx.arc(
        cx,
        cy,
        15,
        0,
        2 * Math.PI
    );

    ctx.fillStyle = "white";
    ctx.fill();


    /* Joint */

    ctx.beginPath();

    ctx.arc(
        x1,
        y1,
        13,
        0,
        2 * Math.PI
    );

    ctx.fillStyle = "white";
    ctx.fill();


    /* End effector */

    ctx.beginPath();

    ctx.arc(
        x2,
        y2,
        10,
        0,
        2 * Math.PI
    );

    ctx.fillStyle = "#ffcc55";
    ctx.fill();


    /* Labels */

    ctx.fillStyle = "white";
    ctx.font = "14px Arial";

    ctx.fillText("BASE", cx + 18, cy);

    ctx.fillText("JOINT 1", x1 + 15, y1);

    ctx.fillText("END", x2 + 15, y2);

    ctx.fillStyle = "#ff5577";

    ctx.fillText(
        "TARGET",
        tx + 15,
        ty
    );
}


function animate() {

    cancelAnimationFrame(animation);


    function frame() {

        currentT1 +=
            (finalT1 - currentT1) * 0.08;


        currentT2 +=
            (finalT2 - currentT2) * 0.08;


        drawRobot(
            currentT1,
            currentT2
        );


        if (
            Math.abs(finalT1 - currentT1) > 0.0001 ||
            Math.abs(finalT2 - currentT2) > 0.0001
        ) {

            animation =
                requestAnimationFrame(frame);
        }
    }


    frame();
}


function resetRobot() {

    document.getElementById("L1").value = 5;
    document.getElementById("L2").value = 3;

    document.getElementById("targetX").value = 6;
    document.getElementById("targetY").value = 2;

    document.getElementById("mode").value = "up";

    currentT1 = 0;
    currentT2 = 0;

    finalT1 = 0;
    finalT2 = 0;

    document.getElementById("theta1").innerText = "--";
    document.getElementById("theta2").innerText = "--";
    document.getElementById("error").innerText = "--";

    document.getElementById("status").innerText =
        "Enter values and calculate";

    drawRobot(0, 0);
}

</script>

</body>
</html>
'''

display(HTML(html_code))
```
